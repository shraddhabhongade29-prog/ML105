# Experiment: Train/Test Split and K-Fold Cross-Validation on Iris



## Aim
To apply train/test split and 7-fold cross-validation on the Iris dataset and compare Accuracy, Precision, Recall and F1-score.

**Author: Dr. Priya B. Pandharbale**

## Step 1: Import Libraries

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_validate, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## Step 2: Load the Iris Dataset

```python
iris = load_iris()
X = iris.data
y = iris.target

print("Total records:", len(X))
print("Classes:", iris.target_names)
```

## Step 3: Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.7, random_state=47, stratify=y)

model = DecisionTreeClassifier(random_state=47)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average="macro"))
print("Recall:", recall_score(y_test, y_pred, average="macro"))
print("F1-score:", f1_score(y_test, y_pred, average="macro"))
```

## Step 4: Apply 7-Fold Cross-Validation

The dataset is divided into seven parts. The model is trained seven times, each time using a different part for validation.

```python
cv = StratifiedKFold(n_splits=7, shuffle=True, random_state=47)

scores = cross_validate(
    DecisionTreeClassifier(random_state=47),
    X, y, cv=cv,
    scoring=["accuracy", "precision_macro", "recall_macro", "f1_macro"]
)

print("Average Accuracy:", scores["test_accuracy"].mean())
print("Average Precision:", scores["test_precision_macro"].mean())
print("Average Recall:", scores["test_recall_macro"].mean())
print("Average F1-score:", scores["test_f1_macro"].mean())
```

## Step 5: Compare the Results

```python
print("TRAIN/TEST SPLIT")
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average="macro"))
print("Recall   :", recall_score(y_test, y_pred, average="macro"))
print("F1-score :", f1_score(y_test, y_pred, average="macro"))

print("\n5-FOLD CROSS-VALIDATION (AVERAGE)")
print("Accuracy :", scores["test_accuracy"].mean())
print("Precision:", scores["test_precision_macro"].mean())
print("Recall   :", scores["test_recall_macro"].mean())
print("F1-score :", scores["test_f1_macro"].mean())
```

## Observations

1. Train/test split evaluates the model using one separate test set.
2. In 7-fold cross-validation, the model is trained and validated seven times.
3. Accuracy measures the proportion of correctly classified samples.
4. Precision measures how many predicted class labels are correct.
5. Recall measures how many actual samples are correctly identified.
6. F1-score is the balance between precision and recall.
7. Macro averaging gives equal importance to each Iris class.

## Result

Train/test split and 7-fold cross-validation were applied to the Iris dataset using a Decision Tree classifier. Accuracy, Precision, Recall and F1-score were calculated and compared.
